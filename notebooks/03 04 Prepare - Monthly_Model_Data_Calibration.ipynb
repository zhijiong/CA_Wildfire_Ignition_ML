# Prepare Monthly Model Data — Calibration Split

Split monthly model data into a **10% calibration** set and a **90% non-calibration** set, stratified by `IS_FIRE`.

## Setup

```python
import os
import pandas as pd
import numpy as np
import gc
from sklearn.model_selection import train_test_split

gc.collect()
```

## Load Data

```python
PROJECT_ROOT = r"E:\zcao\CA_Wildfire"
data_version = 'Monthly_03152026'
input_path   = os.path.join(
    PROJECT_ROOT, 'Clean_Data', 'Model_Data',
    'Evaluation', 'Features_w_Label', data_version
)

mod_Human = pd.read_parquet(os.path.join(input_path, 'monthly_model_data_by_year.parquet'))
print(f'Loaded shape: {mod_Human.shape}')
print(f'Fire rate   : {mod_Human["IS_FIRE"].mean():.4f}')
```

## Calibration Split (10% calibration / 90% non-calibration)

Stratified by `IS_FIRE` to preserve fire rate in both splits.

```python
val_data, non_val_data = train_test_split(
    mod_Human, test_size=0.9, stratify=mod_Human['IS_FIRE'], random_state=315
)

print(f'Calibration set     : {val_data.shape}   fire rate: {val_data["IS_FIRE"].mean():.4f}')
print(f'Non-calibration set : {non_val_data.shape}  fire rate: {non_val_data["IS_FIRE"].mean():.4f}')
```

## Save

```python
data_version = 'Monthly_03152026_calibration'
save_path    = os.path.join(
    PROJECT_ROOT, 'Clean_Data', 'Model_Data',
    'Evaluation', 'Features_w_Label', data_version
)
os.makedirs(save_path, exist_ok=True)
```

```python
cal_path     = os.path.join(save_path, 'monthly_model_data_by_year_calibration.parquet')
non_cal_path = os.path.join(save_path, 'monthly_model_data_by_year_non_calibration.parquet')

val_data.to_parquet(cal_path, index=False)
non_val_data.to_parquet(non_cal_path, index=False)

print(f'Saved calibration     -> {cal_path}')
print(f'Saved non-calibration -> {non_cal_path}')
```
